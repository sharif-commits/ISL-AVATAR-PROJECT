```python
import cv2
import os

video_path = "Apple.mp4"
output_dir = "Apple"

os.makedirs(output_dir, exist_ok=True)

cap = cv2.VideoCapture(video_path)

frame_id = 1

while cap.isOpened():

    ret, frame = cap.read()

    if not ret:
        break

    filename = os.path.join(output_dir, f"{frame_id:06d}.jpg")
    cv2.imwrite(filename, frame)

    frame_id += 1

cap.release()

print(f"Extracted {frame_id-1} frames.")
```
